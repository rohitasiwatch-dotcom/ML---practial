```python
from sklearn.cluster import DBSCAN
import matplotlib.pyplot as plt

# Data: [Annual Spending, Number of Purchases]
X = [[20, 5],
     [22, 6],
     [25, 7],
     [60, 18],
     [65, 20],
     [70, 22],
     [150, 50]]   # Unusual customer

# Apply DBSCAN
dbscan = DBSCAN(eps=12, min_samples=2)

# Find clusters
labels = dbscan.fit_predict(X)

# Print results
print("Cluster Labels:", labels)

# Plot customers
plt.scatter([x[0] for x in X],
            [x[1] for x in X],
            c=labels)

plt.title("Customer Shopping Behavior using DBSCAN")
plt.xlabel("Annual Spending (Thousands)")
plt.ylabel("Number of Purchases")
plt.show()
```
